13. Data Visualization

This section presents visual representations of the key findings from the employee dataset using Matplotlib and Seaborn. The visualizations help identify patterns and trends related to employee distribution, work mode, attrition, performance, salary, and job satisfaction.

13.1 Employee Count by Department

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Department")

plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
```

13.2 Employee Distribution by Work Mode

```python
import matplotlib.pyplot as plt
import seaborn as sns
work_mode_counts = df["Work Mode"].value_counts()
plt.figure(figsize=(7,7))
plt.pie(work_mode_counts,labels=work_mode_counts.index,autopct="%.2f%%")
plt.title("Employee Distribution by Work Mode")
plt.show()
```

13.3 Attrition by Department

```python
import matplotlib.pyplot as plt
import seaborn as sns
attrition_df = df[df["Attrition"] == "Yes"]
plt.figure(figsize=(8,5))
sns.countplot(x="Department", data=attrition_df)
plt.title("Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

13.4 Attrition by Gender

```python
import matplotlib.pyplot as plt
import seaborn as sns
attrition_df = df[df["Attrition"] == "Yes"]
plt.figure(figsize=(7,5))
sns.countplot(data=attrition_df, x="Gender")
plt.title("Attrition by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Employees")
plt.show()
```

13.5 Performance Rating Distribution

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.countplot(x="Performance Rating",data=df)
plt.title("Performance Rating Distribution")
plt.xlabel("Performance Rating")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.show()
```

13.6 Average Salary by Department

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.barplot(x="Department",y="Salary",data=df)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.show()
```

13.7 Salary Distribution by Performance Rating

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.boxplot(data=df,x="Performance Rating",y="Salary")
plt.title("Salary Distribution by Performance Rating")
plt.xlabel("Performance Rating")
plt.ylabel("Salary")
plt.show()
```

13.8 Job Satisfaction Distribution

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.countplot(data=df,x="Job Satisfaction")
plt.title("Job Satisfaction Distribution")
plt.xlabel("Job Satisfaction")
plt.ylabel("Number of Employees")
plt.show()
```

14. Business Insights

14.1 Attrition by Department

The Operations department has the highest employee attrition, indicating a potential employee retention concern. Management should investigate factors such as workload, compensation, job satisfaction, and working conditions to understand why employees are leaving.

14.2 Attrition by Gender

Male employees have higher attrition than female employees in the analyzed dataset. The organization should investigate factors such as salary, job role, department, and job satisfaction to understand the reasons behind this difference and identify appropriate retention strategies.

14.3 Overall Employee Attrition

The overall employee attrition rate is 26.53%, indicating a potential employee retention concern. The organization should analyze factors such as salary, job role, department, and job satisfaction to identify possible reasons for employee turnover and develop effective retention strategies.

14.4 Job Satisfaction and Attrition

Employees with a job satisfaction level of 3 have the highest attrition rate. The organization should investigate factors such as department, salary, performance, work mode, and job role to identify possible reasons for higher employee turnover and improve employee retention.

14.5 Remote Work

Remote work has the highest number of employees in the organization. Management should evaluate employee productivity and satisfaction, ensure that remote employees have the necessary resources, and assess communication, collaboration, and the effectiveness of current remote-working policies.

14.6 Performance and Employee Retention

Performance Rating 4.4 is the most common rating among employees. Management should evaluate whether high-performing employees are receiving appropriate compensation and recognition, while also monitoring their job satisfaction and attrition to support effective employee retention.
